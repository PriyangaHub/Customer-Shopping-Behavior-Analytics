```python
subscription_summary = df_cleaned.groupby("Subscription Status").agg(
    Total_Records=("Customer ID", "count"),
    Average_Purchase=("Purchase Amount (USD)", "mean"),
    Total_Purchase=("Purchase Amount (USD)", "sum")
)

subscription_summary
```

```python

import matplotlib.pyplot as plt

subscription_summary["Average_Purchase"].plot(
    kind="bar",
    title="Average Purchase Amount by Subscription Status",
    xlabel="Subscription Status",
    ylabel="Average Purchase Amount (USD)"
)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Re-create the cleaned DataFrame from the original data
df_cleaned = df.copy()

# Fill missing ratings using the median of each product category
df_cleaned["Review Rating"] = df_cleaned["Review Rating"].fillna(
    df_cleaned.groupby("Category")["Review Rating"].transform("median")
)

# Check the result
print(df_cleaned["Review Rating"].isnull().sum())
```

```python

import re

df_cleaned.columns = [
    re.sub(r'[^a-zA-Z0-9]+', '_', column).strip('_').lower()
    for column in df_cleaned.columns
]

df_cleaned.columns.tolist()
```
